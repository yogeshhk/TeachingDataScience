```python
import outetts
from IPython.display import Audio, display
```

```python
model_config = outetts.HFModelConfig_v2(
    model_path="OuteAI/OuteTTS-0.3-1B", tokenizer_path="OuteAI/OuteTTS-0.3-1B"
)

interface = outetts.InterfaceHF(model_version="0.3", cfg=model_config)
```

```python
interface.print_default_speakers()
```

```python
speaker = interface.load_default_speaker(name="en_male_1")
```

```python
%%time

text = """
Hi everyone, my name is Venelin
and in this video I'll show you a new
free TTS model that can also do voice cloning
""".strip()

gen_cfg = outetts.GenerationConfig(
    text=text,
    temperature=0.4,
    repetition_penalty=1.1,
    max_length=4096,
    speaker=speaker,
)
output = interface.generate(config=gen_cfg)
output.save("intro.wav")
display(Audio("intro.wav"))
```

```python
def generate_speech(
    text: str,
    speaker,
    save_file: str,
    temperature: float = 0.4,
    repetition_penalty: float = 1.1,
):
    gen_cfg = outetts.GenerationConfig(
        text=text,
        temperature=temperature,
        repetition_penalty=repetition_penalty,
        max_length=4096,
        speaker=speaker,
    )
    output = interface.generate(config=gen_cfg)
    output.save(save_file)
```

```python
lyrics = """
I'll take you to the candy shop
I'll let you lick the lollipop
Go 'head, girl, don't you stop
Keep goin' until you hit the spot, whoa
""".strip()
```

```python
%%time

speaker = interface.load_default_speaker(name="en_male_1")
generate_speech(lyrics, speaker, "hip-hop.wav", temperature=0.6)
display(Audio("hip-hop.wav"))
```

## Multilingual

```python
text = """
La science n'est pas simplement une accumulation de faits,
mais un effort pour relier des idées et découvrir des lois cachées
derrière le chaos apparent. La créativité et l'intuition sont essentielles pour
dépasser les limites du raisonnement mécanique, car les vérités profondes
émergent souvent de la contemplation et de la réflexion.
""".strip()
```

## Voice cloning

```python
speaker = interface.create_speaker(audio_path="dwight-voice.mp3", transcript=None)
```

```python
%%time

text = """
Hi everyone, my name is Venelin
and in this video I'll show you a new
free TTS model that can also do voice cloning
""".strip()

gen_cfg = outetts.GenerationConfig(
    text=text,
    temperature=0.5,
    repetition_penalty=1.1,
    max_length=4096,
    speaker=speaker,
)
output = interface.generate(config=gen_cfg)
output.save("voice-clone.wav")
display(Audio("voice-clone.wav"))
```
